# Data manipulation with Pandas

Pandas is a library specialized in the management and analysis of data structures in Python. It was created in 2008 and in 2009 it became an open source platform.

Features of [Pandas](https://pandas.pydata.org/about/):

- Create easily manipulable and indexed ```DataFrame``` objects.
- Reading and writing data in different formats: csv, txt, xlsx, SQL, HDF5.
- Easy handling of objects with missing data.
- Ease of applying ```reshape```.
- Intelligent slicing, easy indexing, and convenience in selecting subsets of ```DataFrames```.
- Easy handling of time series (signals).
- Extensive use in academic and commercial applications, e.g., finance, neuroscience, economics, statistics, advertising, web analytics, etc.

**Pandas DataFrames vs. NumPy Arrays**

- Pandas works with tabular data; NumPy with numerical data.
- Pandas performs better for more than 500k rows, while NumPy performs better for less than 50k rows.
- Pandas consumes more memory than NumPy.
- A DataFrame is a 2D table, whereas an array is a multidimensional object.
- Both Pandas and NumPy are widely used in the [industry](https://www.javatpoint.com/pandas-vs-numpy#:~:text=NumPy%20library%20provides%20objects%20for,memory%20as%20compared%20to%20Pandas.) (Pandas is mentioned in 73 industrial consortia, NumPy in 62.)

**DataFrames creation**

Unlike NumPy arrays that were based on lists, DataFrames are based on dictionaries.

An empty DataFrame is created as follows:

```python
df = pd.DataFrame()
```

The general syntax for creating DataFrames is:
```python
df = pd.DataFrame({'column1': [column1 data],
                   'column2': [column2 data],
                   ...,
                   index=[row names - optional]})
```

If you want to add a column:

```python
df['New Column Name'] = [data]  # It must have the same number of rows as the other columns
```

**Indexing and Slicing in DataFrames**

To see all the data in a column, it's done similar to accessing a dictionary:

```python
df['existing column']
```

For slicing, two methods can be used:

- ```loc```: based on the label of rows or columns.
- ```iloc```: based on the position (integer number) of rows or columns. In general, it would be ```df.iloc[row slicing, column slicing]``` (similar to slicing in collections)

**Note about Series**

`Series` are one-dimensional `DataFrames`, meaning they consist of a single column. Unlike conventional DataFrames, where the dictionary keys represent columns, in `Series`, the keys represent indices since there is only one column.

Some [attributes, methods, and functions](https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series) are shared with DataFrames, but others are exclusive to Series (such as `.to_frame`).

**SUMMARY**

DataFrames are very useful for handling large volumes of ordered data. One of the great advantages of Pandas is the large number of DataFrame manipulation methods. Below are several of great utility:

- ```DataFrame.index```: Returns the index (rows) of the DataFrame.
- ```DataFrame.columns```: Returns the columns of the DataFrame.
- ```DataFrame.iloc```: Position-based indexing. The notation is similar to that of a list ([, ]).
- ```DataFrame.loc```: Label-based indexing of rows or columns, or boolean arrays.
- ```DataFrame.shape```: Returns a tuple with the dimensions of the DataFrame.
- ```DataFrame.to_numpy```: Returns a NumPy array with the DataFrame values. Equivalent to ```DataFrame.values```, although the latter will be deprecated.
- ```DataFrame.add_prefix```: Adds a prefix to the column names of the DataFrame. Although it may not seem so, this is very useful when manipulating DataFrames with hundreds or thousands of columns, which can occur in scientific programming. To add suffixes, use ```DataFrame.add_suffix```.
- ```DataFrame.concat```: Concatenates DataFrames along a specific axis (0/rows, 1/columns).
- ```DataFrame.compare```: Compares two DataFrames.

These are just a few examples of the many methods available, which you can find on the official [Pandas](https://pandas.pydata.org/docs/reference/index.html) documentation page.

**Radar plots**

A radar plot, also known as a **spider plot** or a **radar chart**, is a data visualization technique used to display multivariate data in the form of a two-dimensional chart. It is particularly useful for comparing the performance or characteristics of multiple entities across multiple variables or dimensions.

In a radar plot, each variable is represented by a spoke or axis radiating outward from a central point. The values of each variable are plotted along the corresponding axis, and a line or shape connects the data points to form a polygon. The area of the polygon represents the performance or score of the entity across the variables.

Radar plots are commonly used to:

1. Compare the performance or characteristics of multiple entities across multiple dimensions, such as product features, performance metrics, or personality traits.
2. Identify strengths and weaknesses of entities based on their performance across different variables.
3. Visualize patterns or trends in multivariate data.


````python
# Synthetic data

df = pd.DataFrame({
    'group': ['A','B','C','D'],
    'var1': [38, 1.5, 30, 4],
    'var2': [29, 10, 9, 34],
    'var3': [8, 39, 23, 24],
    'var4': [7, 31, 33, 14],
    'var5': [28, 15, 32, 14]})

# ------- PART 1: Create background

# number of variable
categories=list(df)[1:]
N = len(categories)

# What will be the angle of each axis in the plot? (we divide the plot / number of variable)
angles = [n / float(N) * 2 * np.pi for n in range(N)]
angles += angles[:1]

# Initialise the spider plot
ax = plt.subplot(111, polar=True)

# If you want the first axis to be on top:
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)

# Draw one axe per variable + add labels
plt.xticks(angles[:-1], categories)

# Draw ylabels
ax.set_rlabel_position(0)
plt.yticks([10,20,30], ["10","20","30"], color="grey", size=7)
plt.ylim(0,40)


# ------- PART 2: Add plots

# Plot each individual = each line of the data
# I don't make a loop, because plotting more than 3 groups makes the chart unreadable

# Ind1
values=df.loc[0].drop('group').values.flatten().tolist()
values += values[:1]
ax.plot(angles, values, linewidth=1, linestyle='solid', label="group A")
ax.fill(angles, values, 'b', alpha=0.1)

# Ind2
values=df.loc[1].drop('group').values.flatten().tolist()
values += values[:1]
ax.plot(angles, values, linewidth=1, linestyle='solid', label="group B")
ax.fill(angles, values, 'r', alpha=0.1)


plt.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1))

plt.show()
````
