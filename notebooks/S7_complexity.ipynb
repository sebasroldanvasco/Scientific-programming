## Computing complexity & code profiling



```python
# Import the required libraries

import numpy as np
import time
import cProfile
import timeit
from scipy.fft import fft
```

We can use different libraries to perform a code profiling, such as `time`, `cProfile` or `timeit`.

Let's suppose you want to find the mean of a dataset with $N$ data. Perform this operation using `for`loops and vectorization with `numpy`.

```python
x = np.random.rand(100000)

# using for lopps
st = time.time()
N = len(x)
acum = 0
for i in range(N):
  acum += x[i]
x_ = 1/N*acum
print(x_)
et = time.time()
print(f'Total time using loops = {et-st} seconds')

# using vectorization
st = time.time()
print(np.mean(x))
et = time.time()
print(f'Total time using vectorization= {et-st} seconds')
```

Time complexity measures how the runtime of an algorithm increases with the size of the input, typically denoted as $O(n)$, where $n$ is the size of the input. The goal is to approximate the rate at which the time grows, focusing on the largest contributing factors (i.e., ignore constants and lower-order terms).

- Constant time: $O(1)$ — e.g., accessing a specific index in a list.
- Logarithmic time: $O(\log n)$ — e.g., binary search.
- Linear time: $O(n)$ — e.g., iterating through a list.
- Quadratic time: $O(n^2)$ — e.g., nested loops over a list.
- Exponential time: $O(2^n)$ — e.g., recursive solutions with extensive branching.

To find time complexity analytically:

1. Identify loops: for each loop over $n$ items, the time complexity typically adds a factor of $O(n)$
2. Look at recursive calls: if a function calls itself, set up a recurrence relation (e.g., $T(n)=2T(n/2)+O(1)$ for a divide-and-conquer algorithm) and solve it using methods like the [Master theorem](https://en.wikipedia.org/wiki/Master_theorem_(analysis_of_algorithms)) or recursion trees
3. Consider Built-in Operations: know the time complexities of common Python operations, e.g., appending to a list is $O(1)$, sorting a list is $O(n\log n)$, and accessing a dictionary by key is $O(1)$



*   **Example 1**: Finding the maximum of an array



```python
def find_max(arr): # 1 operation
    max_elem = arr[0]  # 1 operation
    for elem in arr:   # n iterations
        if elem > max_elem:  # 1 operation
            max_elem = elem   # 1 operation
    return max_elem    # 1 operation
```

Note that:

Total operations = 2 (initial assignment) + $n$ (iterations) $\times$ (1+1)  + 1 (return variable) = $2n + 3$


We simplify by dropping lower-order terms and constants. So, the complexity is $Ο(n)$

* **Example 2**: Compute the derivative of an array

```python
def compute_derivative(arr):
    n = len(arr)
    derivative = []

    for i in range(1, n):
        derivative.append(arr[i] - arr[i - 1])

    return derivative
```

* `def compute_derivative(arr)`: Time complexity of $O(1)$.

* `n = len(arr)`: Time complexity of $O(1)$.

* `derivative = []`: Time complexity of $O(1)$.

* `for i in range(1, n)`: The loop iterates $n - 1$ times. Therefore, the time complexity of this loop is $O(n)$.

* `derivative.append(arr[i] - arr[i - 1])`: Since appending to a list has an average time complexity of $O(1)$.

* `return derivative`: Time complexity of $O(1)$.

Now, let's sum up the complexities:

Time complexity: $O(1) + O(1) + O(1) + O(1) + O(n)\times O(1) = O(n)$

Space complexity: $O(n)$ (for the derivative list)

So, the overall time complexity of the `compute_derivative` function is $O(n)$, where $n$ is the length of the input array.

* **Example 3**: Compute the derivative of an array but optimized

```python
def compute_derivative(arr):
    n = len(arr)
    if n <= 1:
        return []

    mid = n // 2
    left_derivative = compute_derivative(arr[:mid])
    right_derivative = compute_derivative(arr[mid:])
    merged_derivative = merge_derivatives(left_derivative, right_derivative)

    return merged_derivative

def merge_derivatives(left, right):
    merged = []
    i, j = 0, 0

    while i < len(left) and j < len(right):
        merged.append(right[j] - left[i])
        i += 1
        j += 1

    while i < len(left):
        merged.append(left[i])
        i += 1

    while j < len(right):
        merged.append(right[j])
        j += 1

    return merged
```

1.  `def compute_derivative(arr)`: Time complexity of $O(1)$.

2. Inside the compute_derivative function:

> * `n = len(arr)`:  Time complexity of $O(1)$.

> * `if n <= 1: return []`: Time complexity of $O(1)$.

3. In the main part of the compute_derivative function:

>* `mid = n // 2`: Time complexity of $O(1)$.
>* `left_derivative = compute_derivative(arr[:mid])` and `right_derivative = compute_derivative(arr[mid:])` - Time complexity of $T(n/2)$, according to the Master theorem
>* `merged_derivative = merge_derivatives(left_derivative, right_derivative)`: Time complexity of $O(n)$.

4. `def merge_derivatives(left, right)`: : Time complexity of $O(1)$.

5. Inside the `merge_derivatives` function:

>* The while loops iterate through the left and right derivative arrays once, respectively. Therefore, they take $O(n/2) = O(n)$ time


Now, let's sum up the complexities:

Time complexity: $T(n) = 2 \times T(n/2) + O(n) → O(n \log n)$.

Space complexity: $O(n)$ due to the recursive calls and the additional space required for merging.

### Efficient algorithms

```python
# Example: Sorting a list efficiently using Python's built-in sort method
lst = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]

# Inefficient sorting
sorted_list = sorted(lst)

# Efficient sorting
lst.sort()
print(lst)
```

The main reason `my_list.sort()` is typically more efficient than `sorted(my_list)` is that it doesn't need to create a new list, resulting in better memory usage and potentially faster execution time.

### Loop optimization

```python
# Example: Summing up a list of numbers using a generator expression instead of a loop

numbers = list(range(100000))

# Inefficient loop
st = time.time()
sum_of_numbers = 0
for num in numbers:
    sum_of_numbers += num
et = time.time()
print(f'Elapsed time: {(et-st)*1000} ms')
print("Sum using loop:", sum_of_numbers,'\n')

# One efficient loop
st = time.time()
sum_of_numbers = sum(numbers)
et = time.time()
print(f'Elapsed time: {(et-st)*1000} ms')
print("Sum using a built-in function:", sum_of_numbers,'\n')

# Another Efficient generator expression
st = time.time()
sum_of_numbers = sum(num for num in numbers)
et = time.time()
print(f'Elapsed time: {(et-st)*1000} ms')
print("Sum using generator expression:", sum_of_numbers)
```

### Avoiding duplicate calculations

```python
# Example: Calculating factorials with duplicate calculations

# With duplicate calculations

def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

# Example usage

st = time.time()
print("Factorial of 5 (with duplicate calculations):", factorial(5))
print("Factorial of 100 (with duplicate calculations):", factorial(100))
et = time.time()
print(f'Elapsed time: {(et-st)*1000} ms\n')

# Example: Calculating factorials without duplicate calculations

# Without duplicate calculations
def factorial(n):
    if n == 0:
        return 1
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# Example usage
st = time.time()
print("Factorial of 5 (without duplicate calculations):", factorial(5))
print("Factorial of 100 (without duplicate calculations):", factorial(100))
et = time.time()
print(f'Elapsed time: {(et-st)*1000} ms')
```

## Types of algorithms


### Greedy algorithm

Example: the coin change problem $→$ selects the largest coin denomination at each step until the desired change is obtained.

```python
def coin_change(coins, amount):
    coins.sort(reverse=True)
    num_coins = 0
    for coin in coins:
        num_coins += amount // coin
        print(f'number of coins: {num_coins}')
        amount %= coin
        print(f'New amount: {amount}')
    return num_coins

coins = [1, 2, 5, 10, 20, 50]
amount = 73
print("Minimum number of coins required:", coin_change(coins, amount))
```

### Divide and conquer algorithm

Example: Merge Sort is an algorithm that recursively divides the array into two halves, sorts each half, and then merges them together.

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left_half = merge_sort(arr[:mid])
    right_half = merge_sort(arr[mid:])
    return merge(left_half, right_half)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

arr = [38, 27, 43, 3, 9, 82, 10]
print("Sorted array:", merge_sort(arr))
```

Let compare the efficiency of the Discrete Fourier Trasnform using loops, and using the divide-and-conquer algorithm (Fast Fourier Transform):
$X(k)=\sum_{n=0}^{N-1}x(n)\left(\cos\left(\frac{2\pi kn}{N}\right)-j\sin\left(\frac{2\pi kn}{N}\right)\right)$
for $k=0,1,...,N-1$

```python
x = np.random.rand(1000)

st = time.time()
N = len(x)
dft = np.zeros(N, dtype = 'complex')
for k in range(N):
  for n in range(N):
    dft[k] += x[n]*(np.cos(2*np.pi*k*n/N)-1j*np.sin(2*np.pi*k*n/N))
et = time.time()
elt1 = et-st
print(f'Elapsed time using conventional formula: {elt1*1000} ms')

st = time.time()
Xk = fft(x)
et = time.time()
elt2 = et-st
print(f'Elapsed time a divide and conquer algorithm: {elt2*1000} ms')
print(f'Time ratio = {elt1/elt2} times')
```

### Dynamic programming

Example: Fibonacci sequence.

```python
[0]*10
```

```python
def fibonacci(n):
    fib = [0, 1]
    for i in range(2, n + 1):
        fib.append(fib[i - 1] + fib[i - 2])
    return fib[n]

n = 10
print("Fibonacci number at index", n, "is:", fibonacci(n))
```

### Backtracking algorithm

Example: Subset sum problem.

```python
def subset_sum(nums, target):
    def backtrack(start, path, curr_sum):
        if curr_sum == target:
            subsets.append(path)
            return
        if curr_sum > target:
            return
        for i in range(start, len(nums)):
            backtrack(i + 1, path + [nums[i]], curr_sum + nums[i])

    subsets = []
    backtrack(0, [], 0)
    return subsets

# Example usage:
nums = [2, 3, 5, 7, 8]
target = 10
result = subset_sum(nums, target)
print("Subsets with sum", target, ":", result)
```

### Binary search

Example: To find a target value in a list.

```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

# Example usage:
arr = [2, 3, 4, 10, 40]
target = 10
index = binary_search(arr, target)
if index != -1:
    print("Element", target, "found at index", index)
else:
    print("Element", target, "not found in the array")

```
