### Numpy arrays


Syntax:



A sequence can be a list (1D arrays), nested lists (2D arrays, 3D, ...), or other arrays.

Any arithmetic operation between arrays of the same size is performed element-wise: `map` is not needed!

***Broadcasting***

In linear algebra, multiplying a matrix or a vector by a scalar, and adding a scalar to the same mathematical object, is a trivial operation: the scalar is multiplied or added to each element:

$\kappa\mathbf{A}=\left[ \begin{matrix}
\kappa a_{11} & \kappa a_{12} & \dots & \kappa a_{1n}\\
\kappa a_{21} & \kappa a_{22} & \dots & \kappa a_{2n}\\
\vdots & \vdots & \ddots & \vdots\\
\kappa a_{m1} & \kappa a_{m2} & \dots & \kappa a_{mn}\\
\end{matrix} \right]$

$\kappa+\mathbf{A}=\left[ \begin{matrix}
\kappa+a_{11} & \kappa+a_{12} & \dots & \kappa+a_{1n}\\
\kappa+a_{21} & \kappa+a_{22} & \dots & \kappa+a_{2n}\\
\vdots & \vdots & \ddots & \vdots\\
\kappa+a_{m1} & \kappa+a_{m2} & \dots & \kappa+a_{mn}\\
\end{matrix} \right]$

However, in programming, it may not necessarily be so straightforward because you would have to create a vector or a matrix with the same dimensions as the vector or matrix being operated on, with the scalar in each of its positions, i.e.:

$\kappa\mathbf{A}=\left[ \begin{matrix}
\kappa  & \kappa & \dots & \kappa \\
\kappa  & \kappa & \dots & \kappa\\
\vdots & \vdots & \ddots & \vdots\\
\kappa  & \kappa & \dots & \kappa\\
\end{matrix} \right] ⊙ \left[ \begin{matrix}
a_{11} & a_{12} & \dots & a_{1n}\\
a_{21} & a_{22} & \dots & a_{2n}\\
\vdots & \vdots & \ddots & \vdots\\
a_{m1} & a_{m2} & \dots & a_{mn}\\
\end{matrix} \right]$


However, ***broadcasting*** avoids such a process. Broadcasting refers to the manipulation of arrays of different shapes in arithmetic operations. Under certain conditions, the smaller array is broadcast across the larger array so that they have compatible dimensions. **Classes for numerical integration**

Numerical methods for integration can be written as follows:

\begin{equation}
∫_a^b f(x)dx≈∑_{i=0}^{n-1}w_if(x_i)
\end{equation}

where $w_i$ are weights and $x_i$ are evaluation points $i=0,1,...,n-1$.



Now, we can define a superclass which performs $∑_{i=0}^{n-1}w_if(x_i)$:

`````python
class Integrator:
  def __init__(self, a, b, n):
    self.a, self.b, self.n = a, b, n
    self.points, self.weights = self.construct_method()

  def construct_method(self):
    raise NotImplementedError('no rule in class %s' %
                              self.__class__.__name__)

  def integrate(self, f):
    s = 0
    for i in range(len(self.weights)):
      s += self.weights[i]*f(self.points[i])
    return s
`````
