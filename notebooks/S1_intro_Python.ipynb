### Data types

There are three basic types of data in Python:


*   Numeric: integers (e.g. 5), floating-point numbers (e.g. 3.14159), complex numbers (e.g. 2+3i)
*   Character strings: texts (e.g. 'Hello world')
*   Boolean values: ```True``` y ```False```

````python
# Integer

a = 5
print(type(a))
# floating-point

pi = 3.14159
print(type(pi))

# complex

c = a+pi*1j
print(type(c))

# character strings

s = 'Hello world'
print(type(s))

# boolean values

b = True
print(type(b))

````


Unlike other programming languages, Python is **strongly typed** language but also **dinamically typed** one, that means:

- Strongly typed: It doesn’t implicitly convert between unrelated types when it doesn’t make sense.

`x = 5 + "3"  # Error: can't add int and str`

- Dynamically typed: You don’t need to declare a variable’s type; it’s determined at runtime and can change while the program runs:



```
x = 5        # int
x = "hello"  # now it's a str
```

**Variable Names and Programming Style**

It's not recommended to use "l" (lowercase ell), "O" (uppercase oh), or "I" (uppercase eye) as single-character variable names. In some fonts, they are easily confused with each other and with 1 (one) or 0 (zero).

The name of a variable, function, etc., must **always** start with a letter. The only special character allowed is the underscore (_).

image.png

[Python](https://peps.python.org/pep-0008/#introduction) recommends:

- Classes: *UpperCamelCase* (e.g. ```class Car```, ```class MoneyCash```, ```class CartesianPoint```)
- Methods: *lowerCamelCase* (e.g. ```execute()```, ```runFast()```, ```setParameters()```, ```predict()```)
- Variables and functions: *lowercase* or *lowercase_separated_by_underscores* (e.g. ```flag```, ```temp_1```, ```speed```).  
Single-character variable names should be avoided unless they are temporary, in which case they are typically used. ```i```, ```j```, ```k```, ```m```, y ```n``` for integers, and ```c```, ```d```, y ```e``` for characters.
- Constants: *CAPITALIZED_WITH_UNDERSCORES* (e.g. ```MAX_TEMPERATURE```, ```N_JOBS```, ```LAYERS```)

## Inheritance

Inheritance is a fundamental concept in object-oriented programming where a new class (called a subclass or derived class) is created based on an existing class (called a superclass or base class).



```
class Superclass:
    def __init__(self, common_attribute):
        """
        Constructor to initialize the superclass attributes.
        """
        self.common_attribute = common_attribute
    
    def common_method(self):
        """
        A method that can be used by all subclasses.
        """
        # Method implementation

class Subclass1(Superclass):
    def __init__(self, common_attribute, unique_attribute1):
        """
        Constructor for Subclass1, initializing both superclass and subclass attributes.
        """
        super().__init__(common_attribute)
        self.unique_attribute1 = unique_attribute1
    
    def unique_method1(self):
        """
        A method unique to Subclass1.
        """
        # Method implementation

class Subclass2(Superclass):
    def __init__(self, common_attribute, unique_attribute2):
        """
        Constructor for Subclass2, initializing both superclass and subclass attributes.
        """
        super().__init__(common_attribute)
        self.unique_attribute2 = unique_attribute2
    
    def unique_method2(self):
        """
        A method unique to Subclass2.
        """
        # Method implementation
```

Even though Python can solve symbolic differentiation (e.g. with `SymPy`),functions that are defined in
an algorithmic way (e.g., solution of another mathematical problem),
or functions with branches, random numbers, etc., pose fundamental
problems to symbolic differentiation, and then numerical differentiation
is required.

We want, for instance, the following behavior with the hypothetical function `Derivative`:


```
def f(x):
  return x**3

>>> dfdx = Derivative(f)
>>> x = 2
>>> dfdx(x)
12.000000992884452
```




We can use the following (but not the best approximation):

\begin{equation}
f'(x)\approx \frac{f(x+h)-f(x)}{h}
\end{equation}

with small $h$, e.g. $1\times 10^{-5}$

We can now check the behavior with $\cos$ and $\sin$

We can also check it with a function, e.g. $f(x) = x^3$

We can implement other approximations of derivatives:

1st order backward diff.: $f'(x)\approx \frac{f(x)-f(x-h)}{h}$

2nd order central diff.: $f'(x)\approx \frac{f(x+h)-f(x-h)}{2h}$

3rd order forward diff.: $f'(x)\approx \frac{1}{h}\left(-\frac{1}{6}f(x+2h)+f(x+h)-\frac{1}{2}f(x)-\frac{1}{3}f(x-h)\right)$

4th order central diff.: $f'(x)\approx \frac{4}{3}\frac{f(x+h)-f(x-h)}{h}-\frac{1}{3}\frac{f(x+2h)-f(x-2h)}{4h}$

6th order central diff.: $f'(x)\approx \frac{3}{2}\frac{f(x+h)-f(x-h)}{h}-\frac{3}{5}\frac{f(x+2h)-f(x-2h)}{4h}+\frac{1}{10}\frac{f(x+3h)-f(x-3h)}{6h}$
